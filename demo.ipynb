## Input Style in Python

#### Taking Input of a 2D array

```python
n_city = int(input('Enter Number of Cities: '))
distance = []

for i in range(n_city):
    path_cost = list(map(int, input(f'Enter Path Cost from {i+1}th city: ').split(' ')))
    distance.append(path_cost)
    
print(distance)
```
